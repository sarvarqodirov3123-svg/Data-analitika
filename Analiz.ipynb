```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Data_FIFA.csv")
df
```

```python
jamolar_gollari = df.groupby('team')['goals'].sum().sort_values(ascending=False)
jamolar_gollari
```

```python
pazitsiyalar_raytingi = df.groupby('position')[['player_rating','pass_accuracy']].mean()
pazitsiyalar_raytingi
```

```python
bosqichlarda_pazitsiya_taqsimoti = pd.crosstab(df["tournament_stage"], df["position"])
bosqichlarda_pazitsiya_taqsimoti
```

```python
jamoa_qiymati = df.groupby('team')['market_value_eur'].mean().sort_values(ascending=False).head(10)
jamoa_qiymati
```

```python
eng_yaxshi_bombardirlar = df.groupby('player_name')[['goals', 'assists']].sum().sort_values('goals', ascending=False).head(10)
eng_yaxshi_bombardirlar

```

```python
plt.figure(figsize=(12,8))
sns.barplot(data=eng_yaxshi_bombardirlar)
```

```python
df['goal_conversion_rate'] = np.where(df['shots'] > 0, df['goals'] / df['shots'], 0)
df
```

```python
zarba_samaradorligi = df.groupby('team')['goal_conversion_rate'].mean().sort_values(ascending=False).head(5)
zarba_samaradorligi
```

```python
COLOR_BLUE = '#4472C4'
COLOR_ORANGE = '#ED7D31'

top10_scorers = (
    df.groupby('player_name')[['goals', 'assists']]
      .sum()
      .sort_values('goals', ascending=False)
      .head(10)
)

x = np.arange(len(top10_scorers))
width = 0.4

fig, ax = plt.subplots(figsize=(14, 6))
bars_goals = ax.bar(x - width/2, top10_scorers['goals'], width, label='goals', color=COLOR_BLUE)
bars_assists = ax.bar(x + width/2, top10_scorers['assists'], width, label='assists', color=COLOR_ORANGE)

ax.bar_label(bars_goals, padding=2, fontsize=9)
ax.bar_label(bars_assists, padding=2, fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(top10_scorers.index, rotation=45, ha='right')
ax.set_title("Eng yaxshi bombardirlar — Top 10 o'yinchi (goals & assists)")
ax.legend()
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
COLOR_BLUE = '#4472C4'

top10_goals_team = (
    df.groupby('team')['goals']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(top10_goals_team.index, top10_goals_team.values, color=COLOR_BLUE)
ax.bar_label(bars, padding=3, fontsize=10)

ax.set_title("Eng ko'p gol urgan 10 ta jamoa")
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
COLOR_BLUE = '#4472C4'    # Defender
COLOR_ORANGE = '#ED7D31'  # Forward
COLOR_GRAY = '#A5A5A5'    # Goalkeeper
COLOR_GOLD = '#FFC000'    # Midfielder

# Bosqich x Pozitsiya bo'yicha o'yinchilar soni
stage_position = pd.crosstab(df['tournament_stage'], df['position'])

# Pozitsiyalar tartibini rang bilan mos qilib belgilaymiz
position_order = ['Defender', 'Forward', 'Goalkeeper', 'Midfielder']
colors = [COLOR_BLUE, COLOR_ORANGE, COLOR_GRAY, COLOR_GOLD]
stage_position = stage_position[position_order]

fig, ax = plt.subplots(figsize=(12, 12))
bottom = None
for pos, color in zip(position_order, colors):
    values = stage_position[pos]
    bars = ax.bar(stage_position.index, values, bottom=bottom, label=pos, color=color)
    bottom = values if bottom is None else bottom + values
    ax.bar_label(bars, label_type='center', fontsize=8, color='white')

ax.set_title("Turnir bosqichlari bo'yicha pozitsiyalar taqsimoti")
ax.set_ylabel("O'yinchilar soni")
ax.set_xlabel('')
plt.xticks(rotation=30, ha='right')
ax.legend(title='Pozitsiya', bbox_to_anchor=(1.02, 1), loc='upper left')
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```

Hammasi exceldagi bilan bir xil. Faqat Excelning "Top 10 Filter" bilan pandasning .head(10) biroz boshqacha ishlaydi. Pandas esa bunga qaramaydi, saralangan ro'yxatdan shunchaki birinchi 10 tasini oladi, qolganini teng bo'lsa ham kesib tashlaydi. Ya'ni bu xato emas — ikkala vositaning teng qiymatlarni qanday hisobga olishidagi tabiiy farqi, xolos. Qiymatlarning o'zi (gollar, reytinglar va boshqalar) ikkalasida ham bir xil chiqadi.
